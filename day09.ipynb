```python
#내가 쓴 코드
def factorial_for():
  a = int(input("정수를 입력하세요"))
  factorial = 1
  for i in range(1, a+1):
    factorial *= i
  return factorial
print (factorial_for())
```

```python
#1)반복문을 활용하여 팩토리얼 구현
def factorial_loop(num):
  result = 1
  for i in range(1, num+1):
    result *= i #result = result * i
  return result
print('5!: ', factorial_loop(5))

#2)재귀함수를 활용한 팩토리얼 표현
def factorial_recursive(num):
  #1. 종료 조건 - num의 값이 1이 되면 1을 반환하고 함수를 종료한다
  if num == 1:
    return 1
  #2. 재귀 조건 - num의 값을 하나씩 감소하며 곱셈을 반복한다.
  return num * factorial_recursive(num-1)

# --> 세줄로 끝난다

#유클리드 방정식
def gcd(a,b):
  #종료 조건: 나누어줄 값(b)이 0이라면 나눌값(a)을 반환하고 유클리드 방정식을 멈춘다.
  if b == 0:
    return a
  #재귀 조건: 나누었을때 나머지 또는 나누어줄 값(b)이 0 이 아니라면 해당 값과 두 정수의 나머지를 재귀함수에 적용한다
  return gcd(b, a%b)
print('24와 18의 최대공약수:', gcd(24,18))

#하노이 탑
def hanoi(n, start, end, sub):
  '''
  n: 원반의 대수 (n의 크기가 클 수록 원반의 크기도 커진다)
  start: 시작 기둥
  end: 목적지 기둥
  sub: 거쳐가는 기눙
  '''

  #종료 조건 - 원반이 1개 남으면 그냥 옮기면 끝
  if n == 1:
    print(f'원반 1dmf {start}번 기둥에서. -> {end}번 기둥으로 이동')
    return
  #step1: 맨 및 원반을 제외한 n-1개를 start에서 sub로 이동(end를 거쳐서)
  hanoi(n-1, start, sub, end)

  #step2: 가장 큰 원반을 원래 목적지(end)로 이동
  print(f'원반{n}을 {start}번 기둥에서 -> {end}번 기둥으로 이동')

  #step3: sub에 위치한 n-1개 원반을 최종 목적지(end)로 이동(start를 거쳐서)
  hanoi(n-1, sub, end, start)

 print('하노이 탑 이동 순서')
  # 3개의 원반을 1번 기둥에서 3번 기둥까지 이동
hanoi(3, start=1, end=3, sub=2)
```
